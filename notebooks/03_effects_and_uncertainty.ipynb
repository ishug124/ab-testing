# 3. Effect and uncertainty

We compare made-up conversion outcomes. Welch's t-test does not require equal group variances.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(100)
n = 5000
control = rng.binomial(1, 0.10, size=n)
treatment = rng.binomial(1, 0.11, size=n)
difference = treatment.mean() - control.mean()
se = np.sqrt(treatment.var(ddof=1)/n + control.var(ddof=1)/n)
interval = (difference - 1.96*se, difference + 1.96*se)
test = stats.ttest_ind(treatment, control, equal_var=False)
print("Difference, percentage points:", difference * 100)
print("Approximate 95% interval:", np.array(interval) * 100)
print("P-value:", test.pvalue)
```

A p-value is not the probability the treatment works. Read the effect size and interval too. The interval here is an approximate large-sample interval.
