# 6. Peeking and many comparisons

With a fixed-end test, plan when it ends. Repeatedly checking and stopping when the result looks positive changes the usual false-alarm behavior. This small simulation compares one final look with four looks. Its output varies and is only a demonstration.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(2026)
simulations = 300
n_each = 2500
looks = [500, 1000, 1750, 2500]
at_end = 0
while_peeking = 0

for _ in range(simulations):
    control = rng.binomial(1, .10, size=n_each)
    treatment = rng.binomial(1, .10, size=n_each)
    if stats.ttest_ind(treatment, control, equal_var=False).pvalue < .05:
        at_end += 1
    p_values = [
        stats.ttest_ind(treatment[:n], control[:n], equal_var=False).pvalue
        for n in looks
    ]
    if min(p_values) < .05:
        while_peeking += 1

print("False alarms at end:", at_end / simulations)
print("False alarms at any look:", while_peeking / simulations)
```

If 20 independent metrics each have a 5% false-alarm chance, the chance of at least one is about 64%. Choose one primary metric before the test; treat other slices as exploratory unless you plan an adjustment.
