# 5. Trust checks and sample ratio mismatch

If the planned split was even, observed group counts should be close to even apart from random variation. A mismatch is a warning to investigate before reading the score.

```python
from scipy import stats

observed = [51200, 49800]
total = sum(observed)
expected = [total / 2, total / 2]
result = stats.chisquare(f_obs=observed, f_exp=expected)
print("Observed counts:", observed)
print("Expected counts:", expected)
print("P-value:", result.pvalue)
```

A small p-value does not diagnose the cause. Check duplicate assignment, redirects, missing events, eligibility rules, and bot filters.
