# 10. Optional: Bayesian uncertainty and bandits

A Bayesian analysis describes uncertainty about possible rates using a model and stated prior information. A bandit shifts traffic toward options that look better while learning.

```python
import numpy as np

rng = np.random.default_rng(42)
control_success, control_n = 100, 1000
treatment_success, treatment_n = 115, 1000
control_draws = rng.beta(1+control_success, 1+control_n-control_success, size=50000)
treatment_draws = rng.beta(1+treatment_success, 1+treatment_n-treatment_success, size=50000)
print("Chance treatment is higher under this model:",
      np.mean(treatment_draws > control_draws))
```

This probability depends on the model and prior; it does not replace checking the experiment and its guardrails.

A simple Thompson-sampling bandit samples from each option's current uncertainty and directs the next person to the higher sample:

```python
rng = np.random.default_rng(2026)
true_rates = [.05, .06]
wins, losses, sent = [0, 0], [0, 0], [0, 0]
for _ in range(2000):
    draws = [rng.beta(wins[a]+1, losses[a]+1) for a in range(2)]
    arm = int(np.argmax(draws))
    result = rng.binomial(1, true_rates[arm])
    sent[arm] += 1
    wins[arm] += result
    losses[arm] += 1-result
print("People sent to each option:", sent)
print("Conversions:", wins)
```

A bandit may improve short-term allocation, while a fixed A/B test is often better when you need a clean causal estimate across several metrics.
