# 8. Optional: designs for connected systems

If one person's treatment changes another person's outcome, individual assignment may be unfair. Cluster tests assign groups together; switchbacks alternate a whole system across time blocks and must consider carryover.

A factorial design tests combinations of changes. For two changes, there are four combinations:

```python
from itertools import product
for a, b in product(["Off", "On"], repeat=2):
    print("Change A:", a, "| Change B:", b)
```

If randomization is impossible, a quasi-experiment such as difference-in-differences may help. It relies on extra assumptions; one important one is that, without treatment, comparison-group trends would have moved similarly.
