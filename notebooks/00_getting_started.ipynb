```python
people = 5
converted = 3
rate = converted / people
print("Conversion rate:", rate)
```

0.6 means 60%. Each 1 in this list means a conversion; 0 means no conversion.

```python
outcomes = [0, 1, 0, 1, 1]
rate = sum(outcomes) / len(outcomes)
print("People:", len(outcomes))
print("Conversions:", sum(outcomes))
print("Rate:", rate)
```
