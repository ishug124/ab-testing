# 2. Metrics

The pooled click rate weights people with many page views more. The average of personal rates gives each person equal weight.

```python
import numpy as np

views = np.array([1, 1, 2, 10, 10])
clicks = np.array([1, 0, 1, 1, 0])
pooled = clicks.sum() / views.sum()
average_person = (clicks / views).mean()
print("Total clicks / total views:", pooled)
print("Average personal rate:", average_person)
```

For a checkout, purchases per assigned person might be the primary metric, payment errors and refunds guardrails, and assignment counts an invariant check.

```python
assigned_people = 1000
purchases = 120
errors = 8
print("Purchase rate:", purchases / assigned_people)
print("Error rate:", errors / assigned_people)
```
