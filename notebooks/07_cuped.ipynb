# 7. Optional: CUPED

CUPED uses a related measure from before the test to reduce predictable differences between people. The treatment cannot have changed a pre-test measure.

```python
import numpy as np

rng = np.random.default_rng(7)
n = 6000
pre = rng.normal(20, 5, size=n)
assigned = rng.choice([0, 1], size=n)
post = 15 + .7*(pre-20) + rng.normal(0, 8, size=n) + assigned
theta = np.cov(pre, post, ddof=1)[0, 1] / np.var(pre, ddof=1)
adjusted = post - theta*(pre - pre.mean())
print("Raw spread:", post.std(ddof=1))
print("Adjusted spread:", adjusted.std(ddof=1))
print("Adjusted effect:", adjusted[assigned == 1].mean() - adjusted[assigned == 0].mean())
```

The gain depends on how well the pre-test measure predicts the outcome. Do not adjust for a behavior the treatment itself could change.
