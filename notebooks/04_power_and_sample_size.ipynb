# 4. Power and sample size

The minimum detectable effect (MDE) is the smallest effect the planned test has a good chance of noticing. This simple planning formula assumes two equal groups and a two-sided comparison of conversion rates.

```python
import math
from scipy import stats

baseline = 0.10
mde = 0.02
alpha = 0.05
power = 0.80
treatment_rate = baseline + mde
z_alpha = stats.norm.ppf(1 - alpha / 2)
z_power = stats.norm.ppf(power)
variance_sum = baseline*(1-baseline) + treatment_rate*(1-treatment_rate)
n_each = math.ceil((z_alpha + z_power)**2 * variance_sum / mde**2)
print("Approximate people per group:", n_each)
```
