# 1. Random assignment

We simulate 10,000 people. Both groups first have the same true conversion chance, so any observed difference comes from ordinary chance.

```python
import numpy as np

rng = np.random.default_rng(42)
n = 10000
group = rng.choice(["Control", "Treatment"], size=n)
chance = np.full(n, 0.10)
converted = rng.binomial(1, chance)

control = converted[group == "Control"]
treatment = converted[group == "Treatment"]
print("Control rate:", control.mean())
print("Treatment rate:", treatment.mean())
```

Random assignment makes groups comparable on average, not identical in every sample. Now add a small treatment effect.

```python
chance = np.where(group == "Control", 0.10, 0.105)
converted = rng.binomial(1, chance)
print("Control:", converted[group == "Control"].mean())
print("Treatment:", converted[group == "Treatment"].mean())
```
