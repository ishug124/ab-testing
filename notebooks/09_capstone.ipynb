# 9. Capstone: test a learner reminder

This uses made-up data. Check the assignment counts, estimate the conversion difference, review uncertainty, and examine the refund guardrail.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(2026)
n = 8000
variant = rng.choice(["Control", "Treatment"], size=n)
chance = np.where(variant == "Control", .10, .112)
converted = rng.binomial(1, chance)
refund_chance = np.where(variant == "Control", .08, .085)
refunded = rng.binomial(1, refund_chance) * converted

c = variant == "Control"
t = variant == "Treatment"
control, treatment = converted[c], converted[t]
difference = treatment.mean() - control.mean()
se = np.sqrt(treatment.var(ddof=1)/len(treatment) + control.var(ddof=1)/len(control))
interval = (difference - 1.96*se, difference + 1.96*se)
test = stats.ttest_ind(treatment, control, equal_var=False)

print("Group counts:", c.sum(), t.sum())
print("Conversion rates:", control.mean(), treatment.mean())
print("Difference in points:", difference * 100)
print("Approximate interval in points:", np.array(interval) * 100)
print("P-value:", test.pvalue)
print("Refunds per assigned person:", refunded[c].mean(), refunded[t].mean())
```
